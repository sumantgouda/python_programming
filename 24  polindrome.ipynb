```python
# 24. Check Palindrome Number Write a program to check if the given number is a palindrome number.
n=int(input("Enter number:"))
num=n
rev=0
while n>0:
    rem=n%10
    rev=rev*10+rem
    n=n//10
if rev==num:
    print(f"{num} is Polindarm number")
else:
    print(f"{num} is not polindram number")
```
